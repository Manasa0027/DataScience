```python
#data cleaning...................................

df.isnull().sum()


df.isnull().sum()/len(df)*100
```

```python
#filling missing values
agemed=df["Age"].median()

df["Age"]=df["Age"].fillna(agemed)
```

```python
df["Age"].isnull().sum()
```

```python
emb=df["Embarked"].mode()

df["Embarked"]=df["Embarked"].fillna(emb)
```

```python
df=df.drop
```

```python
q1=df["Fare"].quantile(0.25)
q3=df["Fare"].quantile(0.75)
IQR=q3-q1
lb_l=q1-1.5*IQR
up_l=q3+1.5*IQR
df=df[(df["Fare"]>=lb_l)&(df["Fare"]<=up_l)]
```

```python
#check unique values
su=df["Sex"].unique()
print(su)
```

```python
df=df.drop(columns=["PassengerId","Name","Ticket"])
```
